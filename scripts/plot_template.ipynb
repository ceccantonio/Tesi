# Template plot generico

Carica dati da file di testo con `np.loadtxt` e fa un plot con barre d'errore.

Formato atteso del file dati (separato da spazi/tab, righe che iniziano con `#` ignorate):

```
x    y    sigma_y
0.0  0.98 0.10
...
```

Modifica solo la cella **CONFIGURAZIONE** per adattarlo ai tuoi dati.

````python
import numpy as np
import matplotlib.pyplot as plt
````

## Configurazione

````python
FILE_DATI = "data/esempio.txt"

# colonne nel file: indice a partire da 0
COL_X, COL_Y, COL_SIGMA_Y = 0, 1, 2

LABEL_X = "x [unità]"
LABEL_Y = "y [unità]"
TITOLO = "Titolo del grafico"
````

## Caricamento dati

````python
x, y, sigma_y = np.loadtxt(
    FILE_DATI, unpack=True, usecols=(COL_X, COL_Y, COL_SIGMA_Y)
)
x, y, sigma_y
````

## Plot

````python
fig, ax = plt.subplots(figsize=(7, 5))
ax.errorbar(x, y, yerr=sigma_y, fmt="o", markersize=4, capsize=3, label="dati")

ax.set_xlabel(LABEL_X)
ax.set_ylabel(LABEL_Y)
ax.set_title(TITOLO)
ax.grid(True, alpha=0.3)
ax.legend()

fig.tight_layout()
# fig.savefig("plot.png", dpi=300)   # decommenta per salvare
plt.show()
````
