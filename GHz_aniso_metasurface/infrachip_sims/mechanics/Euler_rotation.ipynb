```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial.transform import Rotation as R

def plot_axes(ax, R_matrix, label, color=['r', 'g', 'b']):
    """Plots a set of coordinate axes given a rotation matrix"""
    origin = np.zeros(3)
    for i in range(3):
        ax.quiver(*origin, *R_matrix[:, i], color=color[i], linewidth=2, label=f'{label} {"XYZ"[i]}')

def visualize_rotation(euler_angles, order='xyz'):
    """Visualizes a rotated reference frame using Euler angles"""
    fig, axs = plt.subplots(1, 2, figsize=(12, 6), subplot_kw={'projection': '3d'})
    
    # Plot original axes
    identity_matrix = np.eye(3)
    plot_axes(axs[0], identity_matrix, label='Original')
    axs[0].set_xlim([-1, 1])
    axs[0].set_ylim([-1, 1])
    axs[0].set_zlim([-1, 1])
    axs[0].set_xlabel('X')
    axs[0].set_ylabel('Y')
    axs[0].set_zlabel('Z')
    axs[0].set_title('Original Reference Frame')
    axs[0].legend()
    
    # Apply Euler rotation
    rotation = R.from_euler(order, euler_angles, degrees=True)
    rotated_matrix = rotation.as_matrix()
    plot_axes(axs[1], rotated_matrix, label='Rotated', color=['r', 'g', 'b'])
    
    # Formatting rotated plot
    axs[1].set_xlim([-1, 1])
    axs[1].set_ylim([-1, 1])
    axs[1].set_zlim([-1, 1])
    axs[1].set_xlabel('X')
    axs[1].set_ylabel('Y')
    axs[1].set_zlabel('Z')
    axs[1].set_title(f'Rotated Reference Frame: {euler_angles} ({order} order)')
    axs[1].legend()
    
    plt.show()

# Example usage
euler_angles = [0, 90, 0]  # Change these values (in degrees)
order = 'zxz'  # Change the order of rotations
visualize_rotation(euler_angles, order)
```
